**Finishing up Immersed Boundary Methods**

Where immerse boundary methods really succeed *isn't* for something trivial -- i.e., don't use it for the Poission equation on a circle.

But what about your boundaries moving around? A blood cell moving through a fluid? a blood cell with Sickle Cell Aenemia?

The IBM has the advantage of being **scalable with shapes not fixed in place!**

So, IBM makes sense with time changing. 

Now, we get to talk about **problems varying in time.**

**Time-Dependent PDEs.** Our *model* equation will be the heat equation, 
$$\begin{aligned}
&\frac{\partial C}{\partial t} = k \nabla^2 C, \\ &\quad C(t, x), \quad C(0, x) = f(x), \quad C(t, 0) = C(t, 1) = \cdots
\end{aligned}
$$

Where $C$ could be temperature, concentration, random walking (i.e. proabbility).

Suppose you're constrained to a 1D line and you're equally likely to take a step in one direction or the other. (The result doesn't change if $\Delta x$ is uniform or a random variable with distribution.) Also suppose the universe has no bias (e.g., one will walk left just as much as they walk right). 

Then over some finite time period $\Delta t$, you have stumbled (you're drunk in this hypothetical, btw) up with the same probability of moving from your new spot as you had moving from your last spot.

The graph as time goes on would go up and down randomly at these intervals, sometimes having large deviations.

The distribution of probability $P(x,t)$ that's a time-dependent Gaussian.

Over time, you have a higher likelihood of having spread farther. You're always most (predominantly) likely to be at the center, but your likelihood of being at a given place at a given time is flattening. 

Doing some Taylor analysis (?), we get:
$$
\frac{\partial P}{\partial t} = \frac{\Delta x^2}{2 \Delta t} \frac{\partial ^2 P}{\partial x ^2}
$$

Take the limit as $\Delta x \to 0$ and $\Delta t \to 0$.

If they go to $0$ the same, i.e. $\Delta t = \Delta x$, the center fraction is $0$. 

Alternatively, picking their relationship another way makes it go to $\infty$. But if $\Delta t$ is proportional to $\Delta x ^2$, we have the limit equaling $K$.

(Changing our frame of reference without changing the system itself changes the equation?...)

Suppose Brennan starts off with a very 'jaggedy' probability distribution. (He uses a kitty-shaped distribution, spikes for fur and all). This can be $P(0, x)$.

A short time later, it flattens out. Drunk walkers move from their starting location. This can be $P(\text{medium positive number}, x).$

So on and so forth, as it flattens. Our cat melts.

Note: Einstein's most cited paper is the one that noted the relationship between the $\Delta x^2$ and $\Delta t$ in diffusive things. So to measure if something is diffusive, we test to see if it follows this property. Jean - Perie (?) first estimated Avogadro's number by observing pollen grains. (?)

**Let's Solve It.**
$$\begin{aligned}
&\frac{\partial C}{\partial t} = k \nabla^2 C, \\ &\quad C(t, x), \quad C(0, x) = f(x), \quad C(t, 0) = C(t, 1) = \cdots
\end{aligned}
$$

Although this won't be true for real systems, let's have $K = 1$.

Let's have $\nabla^2 = L$, our discretized Laplacian operator.

So,
$$
\frac{\partial C}{\partial t} = LC.
$$

We can completely solve this new ODE.

If we were in 1 dimension, the solution to this would be simple: $C(t, x) = t^{tL} f(x)$. To be in 1 dimension, we would need for $L$ to be a number. It would be the second derivative of $C$. Not in more dimensions...

(For some reason?) We use Taylor expansion to solve for this function, without even needing to approximate.

$$
    e^{tL} = \left[ I + tL + \frac{1}{2}t^2L^2 + \cdots \right]
$$

We have the helpful fact that $L$ is diagonalizable ("D12"), (See LVS), 
$$
L = PDP^\top , \quad D = 
    \begin{bmatrix} \lambda_1 & & \cdots & 0 \\
    0 & \lambda_2 & & 0 \\
    \vdots & & \ddots & \\
    0 & & \cdots & \lambda_n \end{bmatrix}
$$, 
where $P^\top P = I$

Now, we have 
$$
e^{tL} = P\left[I + tD + \frac{1}{2}t^2 D^2 + \cdots \right]
$$

BUT. Computing $L = PDP^\top $ using the QR algorithm **which calculates diagonalization using QR and does not give QR** is $O(n^3)$. If $L$ is sparse, it's $O(n^2)$.


So to take a matrix exponential with known methods is roughly as expensive as a *full eigen decompositon* (which is what $D$ is).

It's numerically impractical, but in terms of analytics, *it is the gold standard.*

Furthermore,

$$
\left[e^{tL}\right] f(x) \approx \int \frac{1}{\sqrt{4 \pi t}} e^{\frac{-(x-y)^2}{et}} f(x) dy
$$,

Where The integrand besides $f(y)$ is the *Heat Equation Greene's Function.* (Also had some "kernel" - esque name?)

$\left[e^{tL}\right]$ is called the *propogator*.

Simpler still, let's study $\frac{\partial C}{\partial t} = \lambda C$.

We have Forward/Backward Euler.


Break up time $[0, T]$ by $\Delta t = T/N$ and solve at discrete time levels.

$$
\frac{\partial C}{\partial t} \approx \frac{C^{n+1} - C^n}{\Delta t} = \lambda C^?.
$$

Forward Euler:
$$
\frac{\partial C}{\partial t} \approx \frac{C^{n+1} - C^n}{\Delta t} = \lambda C^n,
$$

Backward Euler:
$$
\frac{\partial C}{\partial t} \approx \frac{C^{n+1} - C^n}{\Delta t} = \lambda C^{n+1}.

Last time, we solved the discretized heat equation, analytically. Now we're considering Forward and Backward Euler.

$$
C^n_i = C(n + \Delta t i \Delta x).
$$

Forward is cheaper than backward:
$$\begin{aligned}
C^{n+1} = (I+\Delta t L)C^n
\end{aligned}$$

Is cheaper than
$$
(I - \Delta t L)C^{n+1} = C^n, \quad \text{which is a solve.}
$$

**Stability (i.e., accuracy):**

*Premise:* We can *diagonalize* $L = PDP^\top$. $D$ is eigens of $L$. With a finite Laplacian with periodic boundary conditions,
$$ \begin{aligned}
\lambda_{i,j} = -\frac{4}{\Delta x^2}\left( \sin^2 \left( \frac{i \pi \Delta x}{2}\right) + \sin^2 \left( \frac{j \pi \Delta y}{2}\right) \right) \\
\implies \lambda_{\max} = -\frac{8}{\Delta x ^2} \\
\end{aligned}$$
For Forward Euler:
$$\begin{aligned}
C^{n+1} = (I + \Delta t L)C^n \\
P^\top C^{n+1} = P^\top P (I - \Delta t D) P^\top C^n, \quad \text{let } P^\top C^n = y^n: \\
y^{n+1} = (I + \Delta t D)y^n \\
\implies y_i^{n+1} = (1+ \Delta t \lambda_i)y_i^n.
\end{aligned}$$

the true solution is $y_i^n = e^{(n \Delta t \lambda_i)y^0}$. Forward Euler is still a recurrence relation, not an explicit solution like we have. 
Our FE has a true solution,
$$\begin{aligned}
y_i^{n+1} = (1 + \Delta t \lambda_i) y_i^n \\
\implies y_i^n = (1 + \Delta t \lambda_i)^n y_i^0.
\end{aligned}$$

Following the same process for Backward Euler, we have the exact solution
$$
y_i^n = \left( \frac{1}{1 - \Delta t \lambda_i }\right)^n y_i^0.
$$

In summary, we have the exact solution on hand analytically but not computationally, with 2 approximations available computationally and analytically -- and we would maybe compare how the approximation compares to the analytical solution.

Forward Error:
$$
\text{Err}_\text{rel} = |(1+\Delta t \lambda_i)^n - e^{n \Delta t \lambda _i}|, \, \text{Scaled by $|y_0|$.}
$$

Using the Taylor series on the right and Binomial expansion on the left, plus some simplification,
$$
\text{Err}_\text{rel} = \frac{1}{2} \lambda_i^2 n \Delta t e^{\lambda_i n \Delta t} \Delta t + O(\Delta t^2)
$$

If $\lambda_i < 0$, the exact solution decays as $n \to \infty$. Forward Euler won't decay always. Specifically, if $\lambda < -\frac{2}{\Delta t}$, it blows up and oscellates while doing so.

So if $\lambda_i < 0$, we want to make sure that $|1 + \Delta t \lambda_i | < 1$ for FE to work. We can only choose $\Delta t$, and we choose it for the worst case $\lambda_i = \lambda_max = -\frac{8}{\Delta x^2}$.

Then choose $\Delta t$ such that $|1 + \Delta t \frac{8}{\Delta x^2}| < 1$. This yields
$$
\frac{4\Delta t}{\Delta x^2} < 1.
$$
This is *the* fundamental answer for giving decaying solutions from FE when decaying solutions are to be expected.

It's also *the* fundamendal relationship between $\Delta t$ and $\Delta x$ (not caring about the constant) for lots of things, particularly things related to the heat equation.

Backward Error:

The error is the same, but the stability conditions are nicer.

$$\begin{aligned}
\left( \frac{1}{1-\Delta t \lambda_i}\right)^n y^0 \to 0 \\
\implies | \frac{1}{1 - \Delta t \lambda_i}| < 1 \\
\implies \text{The restriction on $\Delta t$ is much easier and in fact negligible. Algebra left to reader.}
\end{aligned}$$

In conlcusion, the explicit method (FE) has harsher constraints on the parameters, but is extremely fast per time step

The implicit method (BE) has few/no restrictions for stability, but is slow per time step.

Choosing is up to the circumstances, how they constrain stability or speed.
